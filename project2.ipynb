```python
from transformers import pipeline
text_gen = pipeline("text-generation",model="Qwen/Qwen2.5-1.5B-Instruct" , device_map = "auto" , torch_dtype = "auto")
```

```python
message = [
    {"role" : "system" , "content" : "You are a useful assistant."},
    {"role" : "user" , "content": "Where is china and what is the population of it"},
]
for i in range (3):
  result = text_gen(message , max_new_tokens = 150 , do_sample = True , temperature = 1.3)
  print(result[-1]["generated_text"][-1]["content"])
```

```python
import time
def ask(question : str , temperature = 0.3):
  message = [
      {"role" : "system" , "content" : "You are a useful assistant."},
      {"role" : "user" , "content": str(question)},
  ]

  result = text_gen(message , max_new_tokens = 150 , do_sample = True , temperature = temperature )

  return result[-1]["generated_text"][-1]["content"]
```

```python
print(ask("Where is china" , 0.7))
```

```python
import time
history = [
    {"role" : "system" , "content" : "You are an useful assistant"},
]
def chat(user_text):
  global history
  history.append({"role" : "user" , "content" : user_text})
  history = history[:1] + history[-8:]
  print(time.time())
  result = text_gen(history, max_new_tokens=150, do_sample=True, temperature=0.3)
  print(time.time())
  reply = result[0]["generated_text"][-1]["content"]
  history.append({"role" : "assistant" , "content" : reply})
  return reply
```
